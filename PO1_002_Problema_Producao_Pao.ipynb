# Problema da Produção (Pão e Pizza)

O objetivo deste problema é determinar a quantidade ideal de produção de Pães e Pizzas para maximizar o lucro, respeitando a disponibilidade dos ingredientes no estoque.

## Tabela de Consumo e Lucro

| Produto | Farinha (un) | Fermento (un) | Manteiga (un) | Ovo (un) | Lucro (R$/un) |
| :--- | :--- | :--- | :--- | :--- | :--- |
| **Pão** | 1 | 2 | 3 | 0 | 0,22 |
| **Pizza** | 3 | 0 | 1 | 1 | 0,20 |
| **Estoque Disponível** | 60 | 30 | 55 | 18 | - |

## Modelo Matemático

**Variáveis de Decisão:**
* $x_1$: Quantidade de Pães a serem produzidos
* $x_2$: Quantidade de Pizzas a serem produzidas

**Função Objetivo (Maximizar Lucro):**
$\max Z = 0.22x_1 + 0.20x_2$

**Restrições (Limites de Estoque):**
* **Farinha:** $1x_1 + 3x_2 \le 60$
* **Fermento:** $2x_1 \le 30$
* **Manteiga:** $3x_1 + 1x_2 \le 55$
* **Ovo:** $1x_2 \le 18$
* **Não-negatividade e Integralidade:** $x_1, x_2 \ge 0$ (e inteiros)

```python
from ortools.linear_solver import pywraplp

def resolver_problema_producao():
    # Instancia o solver SCIP (específico para Programação Inteira)
    solver = pywraplp.Solver.CreateSolver('SCIP')
    if not solver:
        print("Solver não encontrado.")
        return

    # 1. Variáveis de Decisão (Inteiras com limite inferior 0 e superior infinito)
    x1 = solver.IntVar(0, solver.infinity(), 'Pao')
    x2 = solver.IntVar(0, solver.infinity(), 'Pizza')

    # 2. Restrições de Estoque
    # Farinha: 1*x1 + 3*x2 <= 60
    solver.Add(1 * x1 + 3 * x2 <= 60.0, 'Farinha')

    # Fermento: 2*x1 <= 30
    solver.Add(2 * x1 <= 30.0, 'Fermento')

    # Manteiga: 3*x1 + 1*x2 <= 55
    solver.Add(3 * x1 + 1 * x2 <= 55.0, 'Manteiga')

    # Ovo: 1*x2 <= 18
    solver.Add(1 * x2 <= 18.0, 'Ovo')

    # 3. Função Objetivo: Maximizar Lucro
    # Z = 0.22*x1 + 0.20*x2
    solver.Maximize(0.22 * x1 + 0.20 * x2)

    # 4. Resolve o problema
    status = solver.Solve()

    # 5. Exibe os resultados
    if status == pywraplp.Solver.OPTIMAL:
        print('Solução Ótima Encontrada!')
        print(f'Lucro Máximo projetado: R$ {solver.Objective().Value():.2f}\n')

        print('Plano de Produção Ideal:')
        print(f'- {x1.name()}: {int(x1.solution_value())} unidades')
        print(f'- {x2.name()}: {int(x2.solution_value())} unidades')

        # Validando o uso do estoque
        farinha_usada = 1 * x1.solution_value() + 3 * x2.solution_value()
        fermento_usado = 2 * x1.solution_value()
        manteiga_usada = 3 * x1.solution_value() + 1 * x2.solution_value()
        ovo_usado = 1 * x2.solution_value()

        print('\nConsumo do Estoque:')
        print(f'- Farinha: {int(farinha_usada)} / 60 un')
        print(f'- Fermento: {int(fermento_usado)} / 30 un')
        print(f'- Manteiga: {int(manteiga_usada)} / 55 un')
        print(f'- Ovos: {int(ovo_usado)} / 18 un')
    else:
        print('O problema não possui uma solução ótima viável.')

# Executa a função
resolver_problema_producao()
```
